```python
import pandas as pd
import plotly.express as px


# Dataset: Gapminder — GDP, Life Expectancy, Population by Country
# Source: Gapminder Foundation (gapminder.org)

df = px.data.gapminder()
print(f"Loaded: {len(df)} rows")
print(df.head())
```

```python
# explore

print(df.info())
print("Years:", sorted(df['year'].unique()))
print("Continents:", df['continent'].unique())
print(df.describe().round(1))
```

## Task 1 — Scatter: life expectancy change over time

**What to build:** A scatter showing **GDP per capita vs life expectancy** for **two years** (2002 and 2007) to show how both moved — use **colour for year** (just 2 colours), **one continent only**.

Choose any continent except Africa (that was the example). Highlight the change direction.

> 💡 Filter: `df.loc[df['continent'] == 'YOUR_CHOICE']` then filter years


```python
import pandas as pd
import plotly.express as px

# Filter to Americas
americas = df[df['continent'] == 'Americas'].copy()

# Keep only 2002 and 2007
americas = americas[americas['year'].isin([2002, 2007])]

# Convert year to categorical
americas['year'] = americas['year'].astype(str)

# Scatter plot
fig = px.scatter(
    americas,
    x='gdpPercap',
    y='lifeExp',
    color='year',
    hover_name='country',
    log_x=True,
    title='Countries Across the Americas Moved Toward Greater Prosperity and Longer Lives (2002–2007)',
    labels={
        'gdpPercap': 'GDP per Capita (Log Scale)',
        'lifeExp': 'Life Expectancy (Years)',
        'year': 'Year'
    },
    color_discrete_sequence=['#2A9D8F', '#E76F51']
)

# Equal-sized markers
fig.update_traces(
    marker=dict(
        size=13,
        line=dict(width=1, color='white')
    )
)

# Professional styling
fig.update_layout(
    template='plotly_white',
    width=1000,
    height=600,
    title={
        'x': 0.5,
        'font': {'size': 22}
    },
    legend_title='Year'
)

fig.show()
```

## Task 2 — Bubble chart: tell a story

**What to build:** A bubble chart (full 2007 dataset, all countries) where:
- x = GDP per capita (log scale)
- y = life expectancy
- size = population
- colour = ONE continent highlighted (your choice), all others grey
- At least one annotation explaining the highlighted group's story

> This is the grey-and-highlight technique applied to a bubble chart.


```python
# Task 2
# YOUR CODE HERE
import pandas as pd
import plotly.express as px

# Use only 2007 data
df_2007 = df[df['year'] == 2007].copy()

# Highlight Asia, grey out all others
df_2007['highlight'] = df_2007['continent'].apply(
    lambda x: 'Asia' if x == 'Asia' else 'Other Continents'
)

# Bubble chart
fig = px.scatter(
    df_2007,
    x='gdpPercap',
    y='lifeExp',
    size='pop',
    color='highlight',
    hover_name='country',
    log_x=True,
    size_max=50,
    title='Asia Stands Out as a Region Balancing Population Growth and Rising Living Standards (2007)',
    color_discrete_map={
        'Asia': '#E45756',
        'Other Continents': '#C7C7C7'
    },
    labels={
        'gdpPercap': 'GDP per Capita (Log Scale)',
        'lifeExp': 'Life Expectancy (Years)'
    }
)

# Annotation telling the story
fig.add_annotation(
    x=12000,
    y=74,
    text="Many Asian countries combine large populations<br>with improving wealth and longer life expectancy.",
    showarrow=True,
    arrowhead=2
)

fig.update_layout(
    template='plotly_white',
    width=1100,
    height=650,
    title={
        'x': 0.5,
        'font': {'size': 22}
    },
    legend_title_text=''
)

fig.show()
```
